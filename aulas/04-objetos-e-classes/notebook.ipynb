# Em sala, B: objetos que você já usa

Tudo no Python é **objeto**: um valor que carrega os próprios dados e as operações que sabem lidar com eles. Você usa objetos desde o começo do curso. O ponto é a marca.

````python
resultado = "parcialmente procedente"

print(resultado.upper())
print(resultado.count("e"))
print(type(resultado))
````

`upper` e `count` são **métodos** do tipo `str`: funções que moram dentro do tipo e agem sobre aquele texto. Ninguém precisou passar o texto como argumento, porque o método já sabe de quem ele é. Já `print`, `len` e `type` são **funções**: vivem soltas, e o dado entra entre os parênteses.

Um objeto também pode guardar **atributos**, que são dados, e não ações. A data é um bom exemplo.

````python
from datetime import date

julgamento = date(2026, 9, 1)

print(julgamento.year)        # atributo: um dado guardado, sem parênteses
print(julgamento.weekday())   # método: uma ação, com parênteses (0 é segunda-feira)
print(type(julgamento))
````

# Em sala, C: criando um tipo

`str`, `list` e `date` são tipos que alguém escreveu. Você também pode escrever os seus. A **classe** é o molde, e cada **objeto** é uma peça saída dele.

````python
class Cachorro:
    def __init__(self, nome, raca, idade):
        self.nome = nome
        self.raca = raca
        self.idade = idade

    def latir(self):
        return self.nome + ": au au"


rex = Cachorro("Rex", "vira-lata", 3)
toto = Cachorro("Totó", "poodle", 7)

print(rex.nome, rex.idade)
print(toto.nome, toto.idade)
print(rex.latir())
print(toto.latir())
print(type(rex))
````

### Um método sem `self`

A classe abaixo é quase igual à `Cachorro`, com uma diferença no `latir`. Antes de rodar, aposte: o que acontece na última linha?

````python
class CachorroDistraido:
    def __init__(self, nome):
        self.nome = nome

    def latir():
        return "au au"


bidu = CachorroDistraido("Bidu")
bidu.latir()
````

A mensagem diz que `latir` aceita 0 argumentos e recebeu 1. Mas ninguém passou argumento nenhum. Quem passou foi o Python: em `bidu.latir()`, ele entrega o objeto antes do ponto como primeiro argumento do método. É esse o papel do `self`.

Três coisas para ler no código da classe `Cachorro`:

- `__init__` é o método que roda quando o objeto nasce. `Cachorro("Rex", "vira-lata", 3)` chama `__init__`, e os três valores viram os três atributos.
- `self` é o próprio objeto. Dentro da classe, `self.nome` quer dizer "o nome **deste** cachorro". É por isso que `rex.latir()` diz Rex e `toto.latir()` diz Totó, com o mesmo código.
- Todo método declara `self` como primeiro parâmetro, e quem chama não o passa: em `rex.latir()`, o `rex` antes do ponto é o `self`.
