## 4. Resumir: o número plausível e errado

### 4.1 — “qual a taxa de vitória por setor?”

**O que costuma acontecer.** O código mais simples, `groupby("setor")["ganhou_autor"].mean()`, **acerta**: o pandas pula o vazio sozinho. Mas, como a 0.2 leu sem `dtype`, a média sai como objeto, e um `.round(3)` depois dela quebra com `TypeError: Expected numeric dtype`. A “correção” mais comum da IA é `astype(bool)` antes da média, que conta acordo como vitória. Outros erros: “limpar” o vazio com `fillna(False)`, ou definir vitória pela coluna `resultado`.

**Certo.** Transporte aéreo **0,579** e bureau de crédito **0,095** (Encontro 5, exercício 5.3).

**Mostre na tela.**

| O que a IA fez | Transporte aéreo | Bureau de crédito | Geral |
|---|---|---|---|
| deixou os acordos de fora da conta (o certo) | **0,579** | **0,095** | **0,306** |
| contou acordo como derrota (`fillna(False)`) | 0,417 | 0,078 | 0,181 |
| contou acordo como vitória (`astype(bool)`) | 0,697 | 0,255 | 0,588 |
| contou só `PROCEDENTE` como vitória | 0,262 | 0,039 | — |

**Para a conversa.** Nenhuma linha dá mensagem de erro; todas dão um número plausível. É o slide “Pandas e IA” do Encontro 5. Os 2.847 vazios de `ganhou_autor` são 2.697 acordos e 150 sentenças sem resultado.

```python
# astype(float) depois da média é seguro; astype(bool) antes dela é a armadilha.
processos.groupby("setor")["ganhou_autor"].mean().astype(float).sort_values(ascending=False).round(3)
```

### 4.2 — pedido escrito pelos alunos

**O que costuma acontecer.** A lógica costuma vir certa (`dropna`, média, contagem), mas a sintaxe pode vir misturada. Testado no Gemini do Colab em outubro de 2026: ele escreveu `groupby('setor')['ganhou_autor'].agg(taxa_exito=('ganhou_autor', lambda x: x.mean() * 100), numero_casos=('ganhou_autor', 'count'))`, que quebra com `TypeError: func is expected but received tuple in **kwargs`. São duas formas válidas do `agg` fundidas numa que não existe: ou se escolhe a coluna antes e passa só a conta (`['ganhou_autor'].agg(taxa='mean')`), ou não se escolhe e passa a tupla (`.agg(taxa=('ganhou_autor', 'mean'))`). Tirar um dos dois resolve. O `* 100` também transforma a taxa em porcentagem: 57,9 em vez de 0,579.

**Certo.** Transporte aéreo **0,579** sobre **195** casos; bureau de crédito **0,095** sobre **42**.

**Mostre na tela.**

> Calcule, por setor, a taxa de êxito do autor: a proporção de `True` na coluna `ganhou_autor`. Acordos e sentenças sem resultado têm `ganhou_autor` vazio e devem ficar fora da conta. Mostre também quantos casos entram em cada taxa, e ordene da maior para a menor.

**Para a conversa.** É o “princeso” do código: cada pedaço existe e é comum na internet; a combinação não existe. E é o erro do tipo bom: quebra alto, ao contrário do número plausível da 4.1. Embaixo do erro, o Colab mostra o botão **Explicar o erro**: o Gemini lê o erro e corrige. Erro que avisa é fácil; perigoso é o que não avisa.

```python
(processos.dropna(subset=["ganhou_autor"])
    .groupby("setor")["ganhou_autor"].agg(casos="count", taxa="mean")
    .astype({"taxa": float})
    .sort_values("taxa", ascending=False).round(3))
```

## 6. Uma função

### 6.1 — pedido escrito pelos alunos

**O que costuma acontecer.** Com a especificação completa, a IA costuma acertar mais. Mas, testado no Gemini do Colab em outubro de 2026, com um pedido menos preciso, apareceram três problemas em camadas. (1) O aluno chama a função pelo nome da tabela que ela devolve (`comparacao_taxas_por_setor_ordenada(processos)`, quando a função era `calcular_taxas_por_setor`). Se uma célula anterior criou uma tabela com esse nome, o erro é `'DataFrame' object is not callable`; se não, `NameError`. (2) Com o nome certo, quebra com o mesmo `TypeError` do `agg` misturado da 4.2: o modelo repetiu o erro, porque não aprende com a célula anterior. (3) Corrigido o `agg`, a taxa sai certa, mas a coluna de “casos” conta as **vitórias** (113 em transporte aéreo), não os casos com resultado (195). E um comentário afirma que `ganhou_autor` “já é booleano”, o que é falso no notebook dos alunos.

**Certo.** A função abaixo, que passa no teste da 6.2.

**Mostre na tela.**

Mostre a especificação, para eles colarem no pedido:

> Escreva uma função `taxa_de_exito(processos, por="setor")`. Ela recebe a tabela de processos e o nome de uma coluna. Devolve uma tabela com uma linha por grupo, com o grupo no índice, e duas colunas: `casos`, quantas sentenças têm `ganhou_autor` preenchido, e `taxa`, a proporção de `True` entre elas. Acordos e sentenças sem resultado, que têm `ganhou_autor` vazio, ficam fora da conta. A tabela vem ordenada da maior taxa para a menor.

**Para a conversa.** O pedido virou uma especificação: nome, entrada, saída, o que fazer com o vazio, a ordem. Isso é engenharia de contexto. Se aparecer o `'DataFrame' object is not callable`, dois assuntos do Encontro 2 voltam: o nome de dentro da função não existe fora dela, e o notebook guarda as variáveis de células antigas.

```python
def taxa_de_exito(processos, por="setor"):
    validos = processos.dropna(subset=["ganhou_autor"])
    tabela = validos.groupby(por)["ganhou_autor"].agg(casos="count", taxa="mean")
    tabela["taxa"] = tabela["taxa"].astype(float)
    return tabela.sort_values("taxa", ascending=False)
```

### 6.2 — pedido escrito pelos alunos

**O que costuma acontecer.** Se a função errou o vazio, o teste falha com uma mensagem que diz o quê. A mensagem vai de volta para a IA. **Se o teste falhar com a função certa**, a culpa é da tabela: na seção 4 a IA pode ter gravado o `fillna(False)` ou o `astype(bool)` direto em `processos`, e a coluna `ganhou_autor` perdeu os vazios. Rode de novo a 0.2 para recarregar a base.

**Certo.** `Tudo certo: a função bate com o Encontro 5.`

**Mostre na tela.**

Mostre o teste da célula abaixo para eles colarem. Só mostre **depois** que a função estiver escrita: se o teste estiver no notebook antes, ele vira contexto e a IA escreve para passar nele.

**Para a conversa.** Este teste é o mesmo que vai para o repositório, no Codex: a conferência que aqui é na mão vira teste automático.

```python
r = taxa_de_exito(processos)
assert list(r.columns) == ["casos", "taxa"], "as colunas devem ser casos e taxa, nessa ordem"
assert r.index[0] == "Transporte aéreo", "transporte aéreo deveria estar no topo"
assert round(r.loc["Transporte aéreo", "taxa"], 3) == 0.579, "a taxa de transporte aéreo deveria ser 0,579"
assert r.loc["Transporte aéreo", "casos"] == 195, "transporte aéreo deveria ter 195 casos"

r = taxa_de_exito(processos, por="autor_tipo")
assert round(r.loc["PESSOA FÍSICA", "taxa"], 3) == 0.314, "pessoa física deveria ter 0,314"
assert round(r.loc["PESSOA JURÍDICA", "taxa"], 3) == 0.289, "pessoa jurídica deveria ter 0,289"

print("Tudo certo: a função bate com o Encontro 5.")
```
