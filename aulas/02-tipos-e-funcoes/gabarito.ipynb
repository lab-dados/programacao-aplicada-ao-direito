# Parte A. Variáveis e tipos básicos

## A.1 Variáveis e memória

O sinal `=` não é igualdade matemática. Ele **cria uma variável**: o Python
guarda o objeto em algum lugar da memória e faz a variável apontar para ele.

A imagem certa não é a de uma caixa onde o valor mora dentro, e sim a de uma
**etiqueta** colada em um objeto. A diferença aparece na célula abaixo.

```python
valor = 15000.50
copia = valor        # o mesmo objeto, agora com duas variáveis

valor = 20000.00     # 'valor' passa a apontar para outro objeto

print("valor:", valor)
print("copia:", copia)
```

Trocar a etiqueta não mexe no objeto. O que mudou foi só para onde a
variável `valor` aponta; o objeto `15000.50` continua na memória, ainda etiquetado por
`copia`.

É a mesma lógica dos métodos de texto, que aparecem na seção A.3: `.strip()`
devolve um texto novo em vez de alterar o original.

### O tipo é do objeto, não da variável

Uma mesma variável pode apontar para coisas de espécies diferentes ao longo do
notebook.

```python
ano = "2024"
print(ano, type(ano))

ano = 2024
print(ano, type(ano))
```

Por isso a pergunta certa nunca é "que tipo é `ano`?", e sim "que tipo é
`ano` **agora**?".

### Regras e convenções para nomes de variáveis

- Só letras, números e `_`, e não pode começar com número.
- Maiúsculas contam: `Processo` e `processo` são nomes diferentes.
- A convenção é `letras_minusculas_com_underline`.
- Prefira nomes que expliquem o conteúdo. `vc` economiza três segundos hoje e
  custa dez minutos daqui a seis meses; `valor_causa` não.
- Evite nomes que o Python já usa, como `list`, `type`, `str` ou `sum`. Funciona
  na hora e quebra a função de mesmo nome depois.

## A.3 Texto: posições e métodos

Praticamente tudo que sai de um tribunal chega como texto. Cada caractere de um
texto tem uma posição, e você pode pedir um pedaço com colchetes.

O número unificado do CNJ tem o formato `NNNNNNN-DD.AAAA.J.TR.OOOO`.

```python
numero = "1000123-35.2024.8.26.0100"

print(len(numero))        # quantos caracteres
print(numero[0])          # o primeiro
print(numero[:7])         # sequencial
print(numero[11:15])      # ano de ajuizamento
print(numero[16])         # segmento do Judiciário
print(numero[18:20])      # tribunal
print(numero[-4:])        # unidade de origem, contando do fim
```

Duas regras da contagem:

- **Começa em zero.** O primeiro caractere é o de posição `0`.
- **O fim não entra.** `numero[11:15]` pega da posição 11 até a 14. Em
  compensação, `15 - 11` é exatamente o tamanho do pedaço: quatro caracteres.

Números negativos contam do fim para o começo: `numero[-4:]` são os quatro
últimos.

E repare: o resultado continua sendo **texto**. `numero[11:15]` é `'2024'`, com
aspas.

### Métodos de texto

Um método é uma ação que o objeto sabe executar, chamada com ponto e parênteses.
Métodos de texto **devolvem um texto novo** e não alteram o original.

```python
classe = "  PROCEDIMENTO COMUM CÍVEL  "

print(repr(classe.strip()))                      # sem os espaços das pontas
print(classe.strip().title())                    # primeira letra de cada palavra
print(classe.strip().lower())                    # tudo em minúsculas
print(classe.strip().replace("CÍVEL", "CRIMINAL"))
print(classe.strip().startswith("PROCEDIMENTO"))
print("COMUM" in classe)                         # o texto contém isso?

print(repr(classe))                              # o original continua igual
```

Os métodos foram **encadeados**: `classe.strip().title()` funciona porque
`.strip()` devolve um texto, e todo texto tem `.title()`.

Ninguém decora essa lista. Digite `classe.` e pressione `Tab` para ver as
opções, ou use `help(str.replace)`.

## A.4 Conversões de tipo

### Juntando texto e número em uma frase

## B.1 Sua primeira função

Três valores para limpar, com as operações do Exercício 4 repetidas em cada um.
Execute a célula abaixo e depois leia o que você acabou de executar.

```python
bruto = "  R$ 15.000,50  "
valor_1 = float(bruto.strip().replace("R$ ", "").replace(".", "").replace(",", "."))

bruto = "  R$ 2.300,00  "
valor_2 = float(bruto.strip().replace("R$ ", "").replace(".", "").replace(",", "."))

bruto = "  R$ 980,75  "
valor_3 = float(bruto.strip().replace("R$ ", "").replace(".", "").replace(",", "."))

print(valor_1, valor_2, valor_3)
```

Funcionou, e ainda assim é um código ruim. Responda rápido, sem contar nos
dedos: as três linhas de limpeza são idênticas? Se uma delas tivesse um
`.replace` a menos, você teria percebido?

E agora imagine que chegou um valor escrito `R$1.500,00`, sem espaço depois do
`R$`. São três correções, em três lugares. Você vai fazer duas e esquecer a
terceira, o notebook vai continuar rodando, e o valor esquecido vai sair errado
sem avisar.

Uma **função** guarda esse trecho em um lugar só, com um nome.

```python
def limpar_valor(bruto):
    """Converte um valor de tribunal em número. Ex.: '  R$ 15.000,50  ' -> 15000.5"""
    limpo = bruto.strip().replace("R$ ", "").replace(".", "").replace(",", ".")
    return float(limpo)


print(limpar_valor("  R$ 15.000,50  "))
print(limpar_valor("R$ 2.300,00"))
print(limpar_valor("980,75"))
```

As partes da função:

- `def` anuncia a definição, o **nome** é como você vai chamá-la e, entre
  parênteses, vem o **parâmetro** `bruto`, que é o que ela recebe;
- o texto entre três aspas é a **docstring**, que diz o que a função faz;
- as linhas **indentadas** são o corpo. O recuo de quatro espaços é o que marca
  o que está dentro da função, e não é decoração;
- `return` devolve o resultado para quem chamou.

E repare na ordem das coisas: executar a célula do `def` **não roda nada**. O
Python só guardou a receita com aquele nome. O corpo só executa quando você
escreve `limpar_valor(...)`.

Se você mudar a função, execute a célula do `def` de novo. Senão o Python
continua usando a versão antiga, que ainda está na memória da sessão.

### `return` e `print` não são a mesma coisa

Esta é a confusão mais comum de quem está começando a escrever funções, e ela
some depois de ver uma vez.

```python
def com_print(valor):
    print(valor * 2)


def com_return(valor):
    return valor * 2


a = com_print(10)
b = com_return(10)

print("a =", a)
print("b =", b)
```

`com_print` escreve na tela e devolve `None`. Ela serve para você olhar.

`com_return` devolve o objeto para o programa, e por isso `b` vale `20` e pode
entrar na conta seguinte.

É o mesmo `None` do começo da aula. Uma função sem `return` sempre devolve
`None`, e isso não é erro: a função roda, você não vê reclamação nenhuma, e o
resultado some. O erro aparece algumas linhas adiante, quando você tenta usar o
que ela não devolveu.

## B.2 Parâmetros, valores padrão e docstring

Um parâmetro pode ter um **valor padrão**, usado quando quem chama a função não
diz nada a respeito.

```python
def calcular_custas(valor, percentual=0.01):
    """Calcula as custas como um percentual do valor da causa.

    valor: valor da causa, em reais
    percentual: fração aplicada, em que 0.01 é 1%
    """
    return round(valor * percentual, 2)


print(calcular_custas(20000.0))                          # usa o padrão
print(calcular_custas(20000.0, 0.02))                    # por posição
print(calcular_custas(valor=20000.0, percentual=0.02))   # por nome
```

As variáveis da definição (`valor`, `percentual`) são os **parâmetros**. Os
objetos que você entrega na hora de chamar são os **argumentos**.

Passar por nome é mais longo, mais claro e imune à ordem. Foi o que o Encontro 1
usou em `idioma="pt"`.

A docstring vira documentação de verdade: execute a célula abaixo.

```python
help(calcular_custas)
```

### O que acontece dentro fica dentro

A célula abaixo dá erro de propósito.

```python
def limpar(bruto):
    limpo = bruto.strip()
    return limpo


limpar("   um texto com espaços   ")
print(limpo)
```

`limpo` nasce e morre dentro da função. Do lado de fora, essa variável

A regra prática: tudo que a função precisa entra pelos parâmetros, e tudo que ela
produz sai pelo `return`.

## B.3 Uma função de verdade: o dígito do número CNJ

O Extra 2 do Encontro 1 pedia para conferir o dígito verificador de um número
CNJ e avisava que era deliberadamente difícil, "à espera das funções". Chegou a
hora, e são quatro linhas.

O dígito segue a norma ISO 7064, módulo 97 base 10: monte o número **sem o
dígito**, acrescente `00` ao final, e o dígito é `98` menos o resto da divisão
desse número por `97`.

```python
def digito_cnj(numero: str) -> int:
    """Calcula o dígito verificador de um número CNJ."""
    limpo = numero.replace("-", "").replace(".", "")
    sem_digito = limpo[:7] + limpo[9:] + "00"
    return 98 - int(sem_digito) % 97


print(digito_cnj("1017654-66.2021.8.26.0577"))
```

Linha a linha:

1. `numero.replace(...)` tira a pontuação e deixa os 20 dígitos.
2. `limpo[:7]` é o sequencial e `limpo[9:]` é tudo depois do dígito, que ocupa
   as posições 7 e 8. O `+ "00"` é o que a norma manda acrescentar.
3. `int(...)` transforma esse texto em número, `%` pega o resto e `return`
   devolve.

Repare também no cabeçalho: `numero: str` e `-> int` são **anotações de tipo**.
Elas avisam o que entra e o que sai. O Python não confere nada disso; a anotação
serve para pessoas, para o editor e para a IA que vai ler o seu código.

### Exercício 9: escreva e teste

Escreva `confere_cnj(numero)`, que devolve `True` se o dígito informado dentro do
próprio número bate com o calculado por `digito_cnj`, e `False` se não bate.

Dica: o dígito informado está em `numero[8:10]`, como texto.

Depois teste com dois casos:

- `"1017654-66.2021.8.26.0577"`, que está correto, deve dar `True`;
- `"1017654-67.2021.8.26.0577"`, com o dígito adulterado, deve dar `False`.

```python
def confere_cnj(numero: str) -> bool:
    """Diz se o dígito informado no número CNJ bate com o calculado."""
    informado = int(numero[8:10])
    return informado == digito_cnj(numero)


print(confere_cnj("1017654-66.2021.8.26.0577"))
print(confere_cnj("1017654-67.2021.8.26.0577"))
```

### Como conferir que uma função está certa

Foi o que o Exercício 9 pediu, e vale para qualquer função que você escrever:

1. **Um caso que você conhece.** Um número real, cuja resposta dá para ler no
   próprio dado.
2. **Um segundo caso**, de outro tribunal. Um acerto pode ser coincidência; dois

### Extra 1: a máscara do número CNJ

Escreva `formatar_cnj(digitos)`, que recebe os 20 dígitos sem pontuação e
devolve o número no formato `NNNNNNN-DD.AAAA.J.TR.OOOO`.

`formatar_cnj("10176546620218260577")` deve devolver
`'1017654-66.2021.8.26.0577'`.

Dica: cinco fatias e vários `+`.

```python
def formatar_cnj(digitos: str) -> str:
    """Aplica a máscara NNNNNNN-DD.AAAA.J.TR.OOOO a 20 dígitos."""
    return (
        digitos[:7] + "-" + digitos[7:9] + "." + digitos[9:13] + "."
        + digitos[13] + "." + digitos[14:16] + "." + digitos[16:]
    )


print(formatar_cnj("10176546620218260577"))
```

### Extra 2: quanto tempo faz

Usando o módulo `datetime`, que apareceu no Encontro 1, escreva
`dias_desde(ano, mes, dia)`, que devolve quantos dias se passaram desde aquela
data até hoje.

Dica: a subtração de duas datas devolve um objeto com o atributo `.days`.

```python
from datetime import date


def dias_desde(ano: int, mes: int, dia: int) -> int:
    """Devolve quantos dias se passaram desde a data informada até hoje."""
    return (date.today() - date(ano, mes, dia)).days


print(dias_desde(2024, 3, 15))
```

### Extra 3: o tamanho de uma ementa

Escreva `tamanho_ementa(texto)`, que devolve quantas **palavras** tem um texto.

Dica: `.split()` quebra o texto em pedaços separados por espaço, e `len()` conta
quantos pedaços resultaram. Foi o Exercício 6 do Encontro 1, agora embrulhado em
uma função.

```python
def tamanho_ementa(texto: str) -> int:
    """Devolve o número de palavras de um texto."""
    return len(texto.split())


ementa = """RECURSO ESPECIAL. DIREITO À SAÚDE. FORNECIMENTO DE MEDICAMENTO.
OBRIGAÇÃO SOLIDÁRIA DOS ENTES FEDERADOS. RECURSO PROVIDO."""

print(tamanho_ementa(ementa))
```
