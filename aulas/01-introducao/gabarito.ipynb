### Regras e convenções para nomes

- Só letras, números e `_`. Não pode começar com número.
- Python diferencia maiúsculas de minúsculas: `Processo` e `processo` são nomes
  diferentes.
- A convenção é `letras_minusculas_com_underline`.
- Prefira nomes que expliquem o conteúdo. `vc` economiza três segundos hoje e
  custa dez minutos daqui a seis meses; `valor_causa` não.

# Parte B. Atributos, métodos e módulos

## B.1 Métodos: o que o objeto sabe fazer

Um **método** é uma ação que o objeto sabe executar. Ele é chamado com um ponto,
seguido de parênteses.

O ponto significa "deste objeto aqui". Os parênteses significam "execute".


````python
classe = "  Procedimento Comum Cível  "

print(classe.upper())
print(classe.strip())
print(classe.replace("Cível", "Criminal"))
````

### O objeto original não muda

Métodos de texto **devolvem um texto novo**. Eles não alteram o original. Para
guardar o resultado, é preciso dar um nome a ele.

A função `repr()` mostra o texto com as aspas e os espaços visíveis, o que ajuda
a conferir espaços em branco, um problema comum em dados de tribunais.


````python
classe.strip()
print("O original continua:", repr(classe))

classe_limpa = classe.strip()
print("Agora com um nome:  ", repr(classe_limpa))
````

Ninguém memoriza a lista de métodos. Digite `classe.` e pressione `Tab` para ver
a lista, ou use `help(str.replace)` para ler a documentação.

## B.2 Atributos: o que o objeto é

Enquanto o **método** é uma ação (e leva parênteses), o **atributo** é um dado que
o objeto carrega consigo (e não leva parênteses).

A distinção fica clara com datas.


````python
from datetime import date

distribuicao = date(2024, 3, 15)

# Atributos: dados guardados dentro do objeto. Sem parênteses.
print("Ano:", distribuicao.year)
print("Mês:", distribuicao.month)

# Métodos: ações que o objeto executa. Com parênteses.
print("Dia da semana (0 = segunda):", distribuicao.weekday())
print("Formatado:", distribuicao.strftime("%d/%m/%Y"))
````

**A regra de bolso:** se tem parênteses, é método. Se não tem, é atributo.

Um erro clássico do início é escrever `distribuicao.year()`. O Python responde
`TypeError: 'int' object is not callable`, que traduzido significa "você tentou
executar um número".


## B.3 Funções

Uma **função** existe por conta própria e recebe o objeto como argumento, dentro
dos parênteses:

- `len(classe)` é função: o objeto vai dentro dos parênteses.
- `classe.upper()` é método: o objeto vem antes do ponto.


````python
classe = "Procedimento Comum Cível"

print(len(classe))          # quantos caracteres
print(round(15000.567, 2))  # arredonda
print(int("2024"))          # converte texto em número inteiro
````

## C.2 Transcreva

A célula abaixo chama a função `transcricao` do módulo `labdados`. Repare na
estrutura, que é a mesma de tudo o que vimos hoje:

- `labdados` é o **módulo**;
- `transcricao` é a **função** dele;
- `arquivos=`, `local=`, `idioma=` são **argumentos nomeados**, que ajustam o

### Extra 1: decompondo o número CNJ

O número unificado tem 20 dígitos, distribuídos assim:

| Campo | Tamanho | Significado |
|:---|:---|:---|
| `NNNNNNN` | 7 | Número sequencial na origem |
| `DD` | 2 | Dígito verificador |
| `AAAA` | 4 | Ano de ajuizamento |
| `J` | 1 | Segmento do Judiciário |
| `TR` | 2 | Tribunal |
| `OOOO` | 4 | Unidade de origem |

Escreva um código que, a partir do número abaixo, imprima uma frase no formato:

```
Processo 1017654 ajuizado em 2021 na Justiça Estadual, tribunal 26.
```

Dica: `.replace()` remove os separadores e `texto[9:13]` pega os caracteres das
posições 9 a 12. Para juntar texto e valores, use uma *f-string*:
`f"Processo {sequencial} ajuizado em {ano}."`


````python
numero = "1017654-66.2021.8.26.0577"

limpo = numero.replace(".", "").replace("-", "")

sequencial = limpo[0:7]
ano = limpo[9:13]
tribunal = limpo[14:16]

print(f"Processo {sequencial} ajuizado em {ano} na Justiça Estadual, tribunal {tribunal}.")
````

### Extra 2: o dígito verificador

O dígito verificador do número CNJ segue a norma ISO 7064 (módulo 97, base 10).
O cálculo é: monte o número com o sequencial seguido de ano, segmento, tribunal,
origem e `00` ao final; o dígito é `98` menos o resto da divisão desse número por
`97`.

Confira se o dígito do número acima está correto. Use o operador `%`, que devolve
o resto da divisão.

Este exercício é deliberadamente difícil para o primeiro encontro. Ele volta no
Encontro 2, quando tivermos funções.


````python
numero = "1017654-66.2021.8.26.0577"

limpo = numero.replace(".", "").replace("-", "")

sequencial = limpo[0:7]
digito_informado = limpo[7:9]
resto_do_numero = limpo[9:20]

base = int(sequencial + resto_do_numero + "00")
digito_calculado = 98 - (base % 97)

print("Informado:", digito_informado)
print("Calculado:", str(digito_calculado).zfill(2))
print("Confere:", str(digito_calculado).zfill(2) == digito_informado)
````
