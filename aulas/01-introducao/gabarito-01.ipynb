# Parte B. Atributos, métodos e módulos

## B.1 Métodos: o que o objeto sabe fazer

Um **método** é uma ação que o objeto sabe executar. Ele é chamado com um ponto,
seguido de parênteses:

```python
texto.upper()
```

O ponto significa "deste objeto aqui". Os parênteses significam "execute".


````python
classe = "  Procedimento Comum Cível  "

print(classe.upper())
print(classe.lower())
print(classe.strip())
print(classe.replace("Cível", "Criminal"))
````

### Um detalhe importante: o objeto original não muda

Métodos de texto **devolvem um texto novo**. Eles não alteram o original. Para
guardar o resultado, é preciso dar um nome a ele.


````python
print("Antes:", repr(classe))

classe.strip()
print("Depois de chamar strip() sem guardar:", repr(classe))

classe_limpa = classe.strip()
print("Guardando o resultado:", repr(classe_limpa))
````

A função `repr()` mostra o texto com as aspas e os espaços visíveis. É útil
justamente para conferir espaços em branco, um dos problemas mais comuns em dados
de tribunais.

## B.2 Atributos: o que o objeto é

Enquanto o **método** é uma ação (e leva parênteses), o **atributo** é um dado
que o objeto carrega consigo (e não leva parênteses).

A distinção fica clara com datas.


````python
from datetime import date

distribuicao = date(2024, 3, 15)

# Atributos: dados guardados dentro do objeto. Sem parênteses.
print("Ano:", distribuicao.year)
print("Mês:", distribuicao.month)
print("Dia:", distribuicao.day)

# Método: uma ação que o objeto executa. Com parênteses.
print("Dia da semana (0 = segunda):", distribuicao.weekday())
print("Formatado:", distribuicao.strftime("%d/%m/%Y"))
````

**A regra de bolso:** se tem parênteses, é método. Se não tem, é atributo.

Um erro clássico do início é escrever `distribuicao.year()`. O Python responde
`TypeError: 'int' object is not callable`, que traduzido significa "você tentou
executar um número".

# Parte C. Lendo um número CNJ

Chegou a hora de usar o que vimos em um problema real.

Desde 2008, todo processo judicial brasileiro tem um número no formato unificado
definido pela Resolução 65/2008 do CNJ:

```
NNNNNNN-DD.AAAA.J.TR.OOOO
```

| Campo | Tamanho | Significado |
|:---|:---|:---|
| `NNNNNNN` | 7 | Número sequencial do processo na origem |
| `DD` | 2 | Dígito verificador |
| `AAAA` | 4 | Ano de ajuizamento |
| `J` | 1 | Segmento do Judiciário |
| `TR` | 2 | Tribunal |
| `OOOO` | 4 | Unidade de origem (foro, vara, comarca) |

O campo `J` identifica o segmento:

| Código | Segmento |
|:---|:---|
| 1 | Supremo Tribunal Federal |
| 2 | Conselho Nacional de Justiça |
| 3 | Superior Tribunal de Justiça |
| 4 | Justiça Federal |
| 5 | Justiça do Trabalho |
| 6 | Justiça Eleitoral |
| 7 | Justiça Militar da União |
| 8 | Justiça Estadual |

Os números usados neste notebook são fictícios, montados apenas para o
exercício, mas seguem o formato e o dígito verificador da norma.

Este número é a chave que liga bases diferentes: Datajud, sistemas dos tribunais,
planilhas do escritório. Saber decompô-lo é uma das habilidades mais reutilizáveis
do curso.


## C.1 Separando os campos

O método `.split()` quebra um texto em pedaços, usando um separador. O resultado é
uma **lista**, assunto do Encontro 3. Por enquanto, basta saber que os pedaços são
acessados por posição, começando em zero.


````python
numero = "1000123-35.2024.8.26.0100"

partes = numero.split(".")
print(partes)
````

````python
print("Posição 0:", partes[0])
print("Posição 1:", partes[1])
print("Posição 2:", partes[2])
print("Posição 3:", partes[3])
````

Repare que a posição 0 ainda contém o sequencial e o dígito verificador juntos,
`'1000123-35'`, porque o separador entre eles é o hífen, e não o ponto.


````python
sequencial, digito = partes[0].split("-")

print("Sequencial:", sequencial)
print("Dígito verificador:", digito)
print("Ano:", partes[1])
print("Segmento:", partes[2])
print("Tribunal:", partes[3])
print("Origem:", partes[4])
````

## C.2 Fatiando o texto diretamente

Existe um segundo caminho, que não depende dos separadores: pegar pedaços do
texto por posição. Isso se chama **fatiamento** (*slicing*), e usa colchetes com
dois números: onde começa e onde termina.

O primeiro número é incluído, e o segundo não. `texto[0:7]` pega os caracteres das
posições 0, 1, 2, 3, 4, 5 e 6.


````python
numero = "1000123-35.2024.8.26.0100"
so_digitos = numero.replace(".", "").replace("-", "")

print(so_digitos)
print("Sequencial:", so_digitos[0:7])
print("Dígito:     ", so_digitos[7:9])
print("Ano:        ", so_digitos[9:13])
print("Segmento:   ", so_digitos[13:14])
print("Tribunal:   ", so_digitos[14:16])
print("Origem:     ", so_digitos[16:20])
````

## Desafio extra, se sobrar tempo

O dígito verificador de um número CNJ é calculado por um algoritmo (módulo 97,
base 10, da norma ISO 7064). Sem usar nenhuma biblioteca externa, e usando apenas
o que vimos hoje mais o operador `%` (resto da divisão), tente conferir se o
dígito de um número está correto.

O cálculo é: monte o número com o sequencial, o ano, o segmento, o tribunal, a
origem e `00` ao final; o dígito verificador é `98` menos o resto da divisão desse
número por `97`.

Este é um exercício deliberadamente difícil para o primeiro encontro. Se você
travar, tudo bem: ele volta no Encontro 2, quando tivermos funções.


````python
numero = "1017654-66.2021.8.26.0577"

limpo = numero.replace(".", "").replace("-", "")

sequencial = limpo[0:7]
digito_informado = limpo[7:9]
resto_do_numero = limpo[9:20]

# A norma manda reordenar: sequencial + ano/segmento/tribunal/origem + "00"
base = int(sequencial + resto_do_numero + "00")
digito_calculado = 98 - (base % 97)

print("Dígito informado: ", digito_informado)
print("Dígito calculado: ", str(digito_calculado).zfill(2))
print("Confere:", str(digito_calculado).zfill(2) == digito_informado)
````
