### O dicionário: ler, gravar, perguntar

O dicionário é criado entre chaves, com pares `chave: valor`. Ler é
`decisao["tribunal"]`, o que a revisão já fez. Gravar usa o mesmo colchete, do
lado esquerdo do `=`: se a chave não existe, ela é criada; se existe, o valor é
trocado. E `in` pergunta se a **chave** está no dicionário, não o valor.

````python
decisao["juiz"] = "3ª Vara Cível"       # chave nova: cria
print(decisao["juiz"])

decisao["juiz"] = "4ª Vara Cível"       # chave existente: troca o valor
print(decisao["juiz"])

print("juiz" in decisao)
print("relator" in decisao)
````

### Antes do `if`: um sinal guarda, dois sinais perguntam

`resultado = "PROCEDENTE"` guarda o texto na variável e não responde nada.
`resultado == "PROCEDENTE"` pergunta se os dois lados são iguais e devolve
`True` ou `False`. Toda comparação que cabe num `if` devolve um `bool`: igual,
diferente, maior, menor, está dentro, está fora, está ausente. A célula abaixo
faz uma pergunta por linha sobre a `decisao` da revisão. Antes de rodar, diga
qual linha imprime `True`.


````python
resultado = decisao["resultado"]          # um sinal: guarda

print(resultado == "PROCEDENTE")          # dois sinais: pergunta
print(resultado != "PROCEDENTE")
print(decisao["idade_autor"] >= 60)
print(decisao["valor_causa"] < 20000)
print("CDC" in decisao["fundamentos"])
print("CDC" not in decisao["fundamentos"])
print(decisao["idade_autor"] is None)
````
